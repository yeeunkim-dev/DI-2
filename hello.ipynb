```python
# ─────────────────────────────────────────────
# [확인 셀] Git이 설치돼 있는지 확인 (읽기 전용)
#   터미널에서 이미 git을 썼다면 이 셀로도 버전을 확인할 수 있어요.
# ─────────────────────────────────────────────
import shutil

git_path = shutil.which("git")
if git_path:
    print("✅ Git 발견:", git_path)
else:
    print("⚠ Git을 찾지 못했습니다. https://git-scm.com 에서 설치하세요.")
    print("   (Anaconda에 포함돼 있기도 합니다. 터미널에서 'git --version' 으로 확인하세요.)")
```
